1. Find some real dataset. Load it on Colab, Kaggle etc. Try to visualize some of it

*I selected the Iris flower dataset from kaggle.I loaded the CSV file into Google Colab using Pandas and visualized the relationship between petal length and petal width using a scatter plot. Different colours represent different iris species.*

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/drive/MyDrive/Iris.csv")
display(df.head())
display(df.tail())

sns.scatterplot(
    data = df,
    x = "SepalLengthCm",
    y = "SepalWidthCm",
    hue = "Species"
)

plt.title("Sepal Length vs Sepal width")
plt.show()
```
