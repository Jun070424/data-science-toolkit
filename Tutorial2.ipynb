```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2*2 array")
print(array_2*2)
```

```python
random_3x4 = np.random.rand(3,4)

print("This is my random 3x4 array")
print(random_3x4)
```

```python
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of ones")
print(ones_array)

print("Array of zeros")
print(zeros_array)
```

```python
random_6x3 = np.random.rand(6,3)

print("Original array")
print(random_6x3)

transposed_array = random_6x3.T

print("Tranposed array")
print(transposed_array)

print("New shape:", transposed_array.shape)
```

```python
n = 2
p = 3
q = 4

A = np.random.rand(n,p)
B = np.random.rand(p,q)

print("Array A:")
print(A)
print("Shape A: ", A.shape)

print("Array B:")
print(B)
print("Shape B: ",B.shape)

C = np.dot(A,B)

print("Dot product:")
print(C)

print("Result shape:",C.shape)
```
